```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("datafiniti/consumer-reviews-of-amazon-products")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd

print(os.listdir(path))  # see which files are there
df = pd.read_csv(os.path.join(path, "1429_1.csv"))
df.sample(5)  # show 5 random rows
```

```python
# Number of missing values in each column (biggest first)
print(df.isna().sum().sort_values(ascending=False))
 
# What values does the rating column contain? (should be 1 to 5)
print(df["reviews.rating"].value_counts(dropna=False))
 
# Look at a few reviews to get a feeling for the text
print(df["reviews.text"].head(3).tolist())
```

```python
columns_to_keep = ["asins", "name", "reviews.title", "reviews.text",
                   "reviews.rating", "reviews.doRecommend"]
df = df[columns_to_keep].copy()   
 
df = df.rename(columns={
    "name": "product_name",
    "reviews.title": "title",
    "reviews.text": "text",
    "reviews.rating": "rating",
    "reviews.doRecommend": "do_recommend",
})
print(df.head())
```

```python
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")
 
# Only 1, 2, 3, 4, 5 are valid ratings. Anything else becomes missing.
df.loc[~df["rating"].isin([1, 2, 3, 4, 5]), "rating"] = None
print(df["rating"].value_counts(dropna=False))
```

```python
df["do_recommend"] = (
    df["do_recommend"]
    .astype(str)          # make sure everything is text
    .str.strip()          # remove spaces around it
    .str.upper()          # "true" -> "TRUE"
    .map({"TRUE": 1, "FALSE": 0}))
```

```python
df["asins"] = df["asins"].fillna("UNKNOWN").astype(str).str.strip()
df["product_name"] = df["product_name"].fillna("").astype(str).str.strip()
 
print(df.dtypes)
```

```python
# Remove rows that cannot be used
df["text"] = df["text"].astype("string").str.strip()
df = df[df["text"].notna() & (df["text"] != "")]
print("After removing empty text:", len(df))
 
# 5b. A review without a rating has no label -> remove it
df = df[df["rating"].notna()]
df["rating"] = df["rating"].astype(int)   # now safe to make it a whole number
print("After removing missing ratings:", len(df))
```

```python
df = df.drop_duplicates(subset=["asins", "text"])
print("After removing duplicates:", len(df))
```

```python
import re 
import html 
URL_PATTERN = re.compile(r"https?://\S+|www\.\S+")    # web links
HTML_TAG_PATTERN = re.compile(r"<[^>]+>")             # tags like <br>
NOT_ALLOWED = re.compile(r"[^a-z0-9!?.,' ]+")         # keep letters, digits, basic punctuation
REPEATED_CHAR = re.compile(r"(.)\1{2,}")              # 3+ same characters in a row
MANY_SPACES = re.compile(r"\s+")   

def clean_text(text):
    """Clean one review. Each line does one small job."""
    if not isinstance(text, str):              # missing value -> empty string
        return ""
    text = html.unescape(text)                 # "&amp;" -> "&", "&#39;" -> "'"
    text = HTML_TAG_PATTERN.sub(" ", text)     # remove <br>, <p>, ...
    text = text.lower()                        # "Great" and "great" become the same word
    text = URL_PATTERN.sub(" ", text)          # remove links
    text = text.replace("\u2019", "'")         # curly apostrophe -> normal one (don’t -> don't)
    text = NOT_ALLOWED.sub(" ", text)          # remove emojis and special symbols
    text = REPEATED_CHAR.sub(r"\1\1", text)    # "soooo" -> "soo", "!!!!" -> "!!"
    text = MANY_SPACES.sub(" ", text)          # collapse spaces
    return text.strip()

df["title"] = df["title"].fillna("")
df["clean_text"] = df["title"].apply(clean_text) + " . " + df["text"].apply(clean_text)
df["clean_text"] = df["clean_text"].str.strip(" .")   # remove " ." if the title was empty
 
# Remove reviews that are almost empty after cleaning (fewer than 3 words)
df["n_words"] = df["clean_text"].str.split().str.len()
df = df[df["n_words"] >= 3]
print("After removing very short reviews:", len(df))
```

```python
# 1-2 stars = negative, 3 stars = neutral, 4-5 stars = positive
def rating_to_sentiment(r):
    if r <= 2:
        return "negative"
    elif r == 3:
        return "neutral"
    else:
        return "positive"
 
df["sentiment"] = df["rating"].apply(rating_to_sentiment)
 
# Check the balance of the classes. Expect about 93% positive!
print(df["sentiment"].value_counts())
print((df["sentiment"].value_counts(normalize=True) * 100).round(1))
 
df = df.reset_index(drop=True) 
```

```python
from sklearn.model_selection import train_test_split 
X = df["clean_text"]
y = df["sentiment"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,      # 20% for testing
    random_state=42,    # same split every run
    stratify=y,         # keep the same class percentages in train and test
)
```
